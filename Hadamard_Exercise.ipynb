```python
# import all necessary objects and methods for quantum circuits
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit
from qiskit.visualization import plot_histogram
from qiskit_aer import Aer

# define a quantum register with one qubit
q1 =  QuantumRegister(1,"qreg1")

# define a classical register with one bit
c1 = ClassicalRegister(1,"creg1")

# define our quantum circuit for Experiment 1
qc1 = QuantumCircuit(q1,c1)

# apply x-gate to set the qubit to |1>
qc1.x(q1[0])

# apply h-gate (Hadamard: quantum coin-flipping) to the first qubit
qc1.h(q1[0])

# measure the first qubit
qc1.measure(q1,c1)

# draw the circuit
display(qc1.draw(output='mpl'))

# Use Aer's qasm_simulator
simulator = Aer.get_backend('qasm_simulator')

# Execute the circuit on the qasm simulator
job1 = simulator.run(qc1, shots=10000)

# Grab results from the job
result1 = job1.result()

# Returns counts
counts1 = result1.get_counts(qc1)
print("\nTotal counts for Experiment 1 (one Hadamard gate):",counts1)

# Plot a histogram
plot_histogram(counts1)
```

```python
# Install qiskit and qiskit-aer if not already installed
try:
    import qiskit
    import qiskit_aer
except ImportError:
    !pip install qiskit[visualization]
    !pip install qiskit-aer

# import all necessary objects and methods for quantum circuits
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit
from qiskit.visualization import plot_histogram
from qiskit_aer import Aer # Import Aer from qiskit_aer

# define a quantum register with one qubit
q2 =  QuantumRegister(1,"qreg2")

# define a classical register with one bit
# it stores the measurement result of the quantum part
c2 = ClassicalRegister(1,"creg2")

# define our quantum circuit
qc2 = QuantumCircuit(q2,c2)

# apply x-gate to set the qubit to |1> as per the task description
qc2.x(q2[0])

# apply h-gate (Hadamard: quantum coin-flipping) to the first qubit
qc2.h(q2[0])

# apply h-gate (Hadamard: quantum coin-flipping) to the first qubit once more
qc2.h(q2[0])

# measure the first qubit, and store the result in the first classical bit
qc2.measure(q2,c2)

# draw the circuit by using matplotlib
display(qc2.draw(output='mpl'))

# Use Aer's qasm_simulator
simulator = Aer.get_backend('qasm_simulator')

# Execute the circuit on the qasm simulator
job = simulator.run(qc2, shots=10000)

# Grab results from the job
result = job.result()

# Returns counts
counts = result.get_counts(qc2)
print("\nTotal counts for Experiment 2 (two Hadamard gates):",counts)

# Plot a histogram
plot_histogram(counts)
```
